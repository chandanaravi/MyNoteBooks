```python
import nltk
from nltk.stem import PorterStemmer, SnowballStemmer, LancasterStemmer, WordNetLemmatizer
from nltk.tokenize import word_tokenize


#nltk.download('punkt')
#nltk.download('wordnet')
#nltk.download('omw-1.4')

text = "Studies studying studied runs running runner easily fairly happiness better"
words=word_tokenize(text)
print(words)

porter=PorterStemmer()
snowball=SnowballStemmer('english')
lancaster=LancasterStemmer()
lemmatizer=WordNetLemmatizer()
# Process words
print(f"{'Word':<15} {'Porter':<15} {'Snowball':<15} {'Lancaster':<15} {'Lemma':<15}")
print("-" * 75)
for word in words:
    porter_stem = porter.stem(word)
    snowball_stem = snowball.stem(word)
    lancaster_stem = lancaster.stem(word)
    lemma = lemmatizer.lemmatize(word)

    print(f"{word:<15} {porter_stem:<15} {snowball_stem:<15} {lancaster_stem:<15} {lemma:<15}")
```

In Natural Language Processing (NLP), stemming is the process of reducing words to their root (base) form—often by chopping off prefixes/suffixes. There are several types of stemming approaches, mainly based on how aggressively or intelligently they reduce words.

1. Rule-Based Stemming
2. Dictionary-Based Stemming (Lemmatization-like)
3. Statistical Stemming
4. Hybrid Stemming (Combines rule-based + dictionary + statistical approaches.)
5. Language-Specific Stemming
